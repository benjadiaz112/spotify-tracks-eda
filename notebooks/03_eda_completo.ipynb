Este notebook estudia la popularidad, los géneros, el contenido explícito, las características musicales, las correlaciones y los valores extremos del dataset limpio.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))

from src.analysis.eda_complete import run_full_eda
```

## 1. Ejecutar el análisis y generar evidencias

```python
results = run_full_eda(PROJECT_ROOT)
df = results['data']
print(f'Filas analizadas: {len(df):,}')
print(f'Columnas: {df.shape[1]}')
```

## 2. Distribución de popularidad

```python
df['popularity'].describe().round(2)
```

La popularidad promedio es cercana a 33 puntos y la mediana también es 33. Los valores cubren todo el rango entre 0 y 100.

## 3. Popularidad por género

```python
print('Géneros con mayor mediana:')
print(results['genres'].head(10).to_string(index=False))
print('Géneros con menor mediana:')
print(results['genres'].tail(10).sort_values('mediana').to_string(index=False))
```

El género muestra diferencias claras de popularidad. Sin embargo, la limpieza conserva un género por canción, por lo que esta comparación debe interpretarse con cautela.

## 4. Canciones explícitas y no explícitas

```python
results['explicit']
```

Las canciones explícitas tienen un promedio aproximado de 36,9 puntos, frente a 32,9 para las no explícitas. Es una asociación descriptiva y no demuestra que el contenido explícito cause mayor popularidad.

## 5. Canciones más y menos populares

```python
print('Más populares:')
print(results['top_tracks'].to_string(index=False))
print('Menos populares:')
print(results['bottom_tracks'].to_string(index=False))
```

## 6. Relación con características musicales y correlación

```python
results['correlations'].round(4)
```

Las asociaciones lineales individuales son débiles. `loudness` y `danceability` presentan las correlaciones positivas más altas, pero ninguna supera 0,08. Esto sugiere que la popularidad depende de combinaciones, relaciones no lineales y factores no incluidos en el dataset.

## 7. Valores extremos

```python
results['outliers']
```

La regla IQR identifica principalmente valores extremos en volumen y duración. Estos registros no se eliminan automáticamente: pueden representar canciones reales y deberán evaluarse según el modelo elegido.

## 8. Variables relevantes y conclusiones

- `popularity` es la variable objetivo.
- `track_genre` y `explicit` muestran diferencias descriptivas entre grupos.
- `loudness` y `danceability` tienen las asociaciones lineales más altas, aunque débiles.
- `energy`, `acousticness`, `valence`, `tempo` y `duration_ms` deben conservarse para evaluar relaciones combinadas o no lineales.
- Los identificadores y nombres no deben utilizarse directamente como predictores sin una justificación adicional.
- La popularidad parece depender de factores musicales, culturales y comerciales que no están completamente representados en estos datos.
